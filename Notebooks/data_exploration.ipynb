# Correlation Coefficient

A correlation coefficient ranges from -1 to 1 and tells us if there's a linear relationship.
Close to 0 means no linear relationship, and close to either of the bounds means there is a linear relationship.

Let's look at the relationship between 'Price_per_Gal' and 'Total_Price'. 

We know those already have a correlation (and a causation) but let's prove the correlation with math. 

```python

# Calculate correlation between two columns.
'''df['col1'].corr(df['col2'])'''

# To remove the float datatype message, we can also print it with a float type conversion.
'''float(df['col1'].corr(df['col2']))'''
# TODO

```
