# Stripping Whitespaces

'Whitespace' is the empty space around a variable. For example, ' Total' has a leading whitespace. 
Whitespaces can make filtering and searching quite inaccurate, so it's good to clean these when they are found.

```python
# Strip whitespace from variabeles in rows
'''df["column"] = df["column"].str.strip()'''
#TODO

# Strip whitespace from column names
'''df.columns = df.columns.str.strip()'''
#TODO
```
